1. Supervised Learning
In **Supervised Learning**, the algorithm learns from a *labeled* dataset. We provide the model with both the input features ($X$) and the correct output/target ($y$). The ultimate goal is to map the input to the output to predict targets for new, unseen data.

**Key Categories:**
* **Regression:** Predicts continuous numerical values. 
    *   *Real-World Application:* Predicting real estate prices based on square footage ($f(x) = y$).
* **Classification:** Predicts discrete categorical labels.
    *   *Real-World Application:* Classifying a medical tumor as `Malignant (1)` or `Benign (0)`.

2. Unsupervised Learning
In **Unsupervised Learning**, the algorithm is fed with *unlabeled* data. It must independently discover hidden structures, patterns, or relationships within the dataset without explicit instructions.

**Key Categories:**
*   **Clustering:** Grouping similar data points together based on inherent features.
    *   *Real-World Application:* Customer segmentation in e-commerce or grouping daily news articles by topic.

Key Takeaways
* **Supervised Learning** = Labeled Data (Regression & Classification).
* **Unsupervised Learning** = Unlabeled Data (Clustering).
* The choice of algorithm depends entirely on the nature of the data and the business problem being solved.
