## 1. Model Representation
In linear regression with one variable, we represent our model function $f$ as a straight line equation:
$$f_{w,b}(x) = wx + b$$

**Terminology:**
*   **$x$**: Input feature (e.g., Study hours)
*   **$y$**: Target variable (e.g., Exam score)
*   **$w$**: Weight (Parameter)
*   **$b$**: Bias (Parameter)
*   **$m$**: Number of training examples

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Define the Custom Dataset
# x_train: Number of hours studied
# y_train: Exam score achieved
x_train = np.array([2.0, 5.0, 8.0]) 
y_train = np.array([40.0, 70.0, 95.0])

# 2. Calculate the number of training examples (m)
m = x_train.shape[0]

print(f"Input features (Study Hours): {x_train}")
print(f"Target values (Exam Scores): {y_train}")
print(f"Number of training examples (m): {m}")
```

## 2. Visualizing the Data
Before computing the regression model, it is crucial to understand the dataset by visualizing it. We use a scatter plot to observe the relationship between the number of study hours and the resulting exam scores.

```python
# Plot the data points
plt.scatter(x_train, y_train, marker='x', c='r', label='Actual Data')

# Set the title and axis labels
plt.title("Study Hours vs. Exam Scores")
plt.ylabel('Exam Score')
plt.xlabel('Study Hours')
plt.legend()
plt.show()
```

```python
def compute_model_output(x, w, b):
    """
    Computes the prediction of a linear model
    Args:
      x (ndarray (m,)): Data, m examples 
      w,b (scalar)    : model parameters  
    Returns
      f_wb (ndarray (m,)): model prediction
    """
    m = x.shape[0]
    f_wb = np.zeros(m)
    
    for i in range(m):
        f_wb[i] = w * x[i] + b
        
    return f_wb

# Let's test with some initial w and b values
w = 10
b = 10

# Call the function to get predictions
tmp_f_wb = compute_model_output(x_train, w, b)

print(f"Our initial predictions: {tmp_f_wb}")
```

## 4. Plotting the Model Predictions
Now, we will plot our model's predictions (the straight line) alongside the actual data points. This helps us visually evaluate how well our chosen parameters ($w$ and $b$) fit the dataset.

```python
# Plot our model's predictions (The line)
plt.plot(x_train, tmp_f_wb, c='b', label='Our Prediction')

# Plot the actual data points (The scatter crosses)
plt.scatter(x_train, y_train, marker='x', c='r', label='Actual Data')

# Set titles and labels
plt.title("Study Hours vs. Exam Scores")
plt.ylabel('Exam Score')
plt.xlabel('Study Hours')
plt.legend()
plt.show()
```

## 5. Conclusion & The Next Step
As we can see in the graph, our initial prediction line (blue) doesn't perfectly match the actual data points (red crosses). This is because we manually guessed the parameters $w=10$ and $b=10$. 

**The Core Problem of Machine Learning:** How do we automatically find the absolute best values for $w$ and $b$ so that our line perfectly fits the data? 

To solve this, we need a mathematical way to measure our error (distance between the line and the points). This is called the **Cost Function**, and minimizing that error requires an algorithm called **Gradient Descent**.
